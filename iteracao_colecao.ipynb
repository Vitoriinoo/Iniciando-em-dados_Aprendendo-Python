## 1. Iterando sobre Listas

Listas são coleções ordenadas de itens. Podemos iterar sobre elas de várias formas.


```python
# Exemplo 1: Iterar sobre lista de números
# Crie uma lista notas = [8.5, 7.0, 9.5, 6.5, 8.0]
# Use um laço for para iterar sobre cada nota e imprimir
# Depois, calcule a média: crie soma = 0, some todas as notas e divida por len(notas)
# Escreva seu código aqui

notas = [8.5, 7.0, 9.5, 6.5, 8.0]

soma = 0

for numero in notas:
  soma = soma + numero

media = soma / len(notas)

print(media)
```

```python
# Exemplo 2: Iterar com índice usando range()
# Crie uma lista alunos = ["Ana", "Bruno", "Carlos", "Diana"]
# Use range(len(alunos)) com um laço for para iterar pelos índices
# Para cada índice, imprima o número (i + 1) e o nome do aluno
# Escreva seu código aqui

alunos = ["Ana", "Bruno", "Carlos", "Diana"]

for i in range(len(alunos)):
  print(i+1, end = " ")
  print(alunos[i], end = " ")
```

## 2. Iterando sobre Strings

Strings são sequências de caracteres. Podemos iterar sobre cada caractere!


```python
# Exemplo 1: Contar caracteres de uma palavra
# Crie uma variável palavra = "Python" e contador = 0
# Use um laço for para iterar sobre cada caractere da string
# Para cada caractere, incremente o contador e imprima
# Após o laço, imprima o total de caracteres
# Escreva seu código aqui
palavra = "Python"
contador = 0

for letra in palavra:
  contador = contador + 1

print(contador)
```

```python
# Exemplo 2: Contar vogais em uma palavra
# Crie variáveis: palavra = "Programação", vogais = "aeiouAEIOU", contador_vogais = 0
# Use um laço for para iterar sobre cada caractere da palavra
# Dentro do laço, verifique se a letra está em vogais
# Se for vogal, incremente o contador e imprima
# Após o laço, imprima o total de vogais
# Escreva seu código aqui

palavra = "Programação"
vogais = "aeiouAEIOU"
contador_vogais = 0

for letra in palavra:
  if letra in vogais:
    contador_vogais = contador_vogais + 1
    print(letra)

print(contador_vogais)
```

## 3. Iteração com Condições

Podemos combinar laços com estruturas condicionais para criar lógicas mais complexas.


```python
# Exemplo 1: Filtrar notas aprovadas
# Crie uma lista notas = [8.5, 6.0, 9.0, 5.5, 7.5, 4.0]
# Crie listas vazias: notas_aprovadas = [] e notas_reprovadas = []
# Use um laço for para iterar sobre cada nota
# Dentro do laço, use if/else para separar notas >= 7.0 (aprovadas) das demais (reprovadas)
# Use append() para adicionar cada nota à lista correspondente
# Após o laço, imprima as listas e os totais
# Escreva seu código aqui

notas = [8.5, 6.0, 9.0, 5.5, 7.5, 4.0]

notas_aprovadas = []
notas_reprovadas = []

for nota in notas:
  if nota >= 7:
    notas_aprovadas.append(nota)
  else:
    notas_reprovadas.append(nota)

print(f"Notas aprovadas: {notas_aprovadas}")

print(f"Notas reprovadas: {notas_reprovadas}")
```

```python
# Exemplo 2: Encontrar maior e menor valor
# Crie uma lista numeros = [15, 8, 23, 4, 42, 11, 7]
# Verifique se a lista não está vazia (len(numeros) > 0)
# Se não estiver vazia, inicialize maior e menor com o primeiro elemento
# Use um laço for para iterar sobre os números
# Dentro do laço, atualize maior e menor conforme necessário
# Após o laço, imprima a lista, maior, menor e a diferença
# Escreva seu código aqui

numeros = [15, 8, 23, 4, 42, 11, 7]

if (len(numeros) > 0):
  maior = numeros[0]
  menor = numeros[0]

  for numero in numeros:
    if numero > maior:
      maior = numero
    if numero < menor:
      menor = numero

    dif = maior - menor

print(f"Maior: {maior}")
print(f"Menor: {menor}")
print(f"Diferença: {dif}")
```

## 4. Laços Aninhados

Podemos colocar laços dentro de outros laços! Isso é útil para processar estruturas bidimensionais.


```python
# Exemplo 1: Tabuada completa (1 até 5)
# Use um laço for externo com range(1, 6) para cada número da tabuada
# Dentro dele, use um laço for interno com range(1, 11) para os multiplicadores
# Calcule e imprima cada multiplicação
# Escreva seu código aqui

numeros_1_5 = range(1,6)
numeros_1_10 = range(1,11)

for numero in numeros_1_5:
  for n in numeros_1_10:
    multiplicacao = numero * n
    print(f"{numero} x {n} = {multiplicacao}")
```

```python
# Exemplo 2: Padrão de asteriscos
# Use um laço for externo com range(1, 6) para cada linha
# Dentro dele, use um laço for interno com range(linha) para imprimir asteriscos
# Use end="" para imprimir na mesma linha e print() para pular linha após cada linha
# Escreva seu código aqui

for linha in range(1, 6):
  for valor in range(linha):
    print("*", end = " ")
  print()
```
