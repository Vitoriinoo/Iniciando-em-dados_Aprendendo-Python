## **1. Matrizes (listas de listas) — o essencial**

Uma matriz é uma `list` que contém outras `list`. É útil para:

- dados tabulares (linhas/colunas)
- grades (ex.: tabuleiro)

Padrões que você precisa dominar: **criar**, **acessar** (`m[linha][coluna]`) e **percorrer**.


### 1.1 Exemplo único (criar → acessar → percorrer)

````python
# Crie uma matriz (lista de listas) com 3 linhas e 3 colunas
# A matriz deve conter os números de 1 a 9 organizados em 3 linhas
# Acesse o elemento na linha 0, coluna 0 e imprima
# Acesse o elemento na linha 1, coluna 2 e imprima
# Escreva seu código aqui
        # 0 1 2
lista1 = [1,2,3] # 0
lista2 = [4,5,6] # 1
lista3 = [7,8,9] # 2

matriz = [lista1, lista2, lista3]

print(matriz[0][0])
print(matriz[1][2])
````

````python
# Modifique o elemento na linha 0, coluna 1 da matriz para 99
# Imprima a linha 0 completa após a modificação
# Escreva seu código aqui

matriz[0][1] = 99
print(matriz)
````

````python
# Percorra a matriz usando dois laços for aninhados com enumerate
# O laço externo deve iterar sobre as linhas (com índice i)
# O laço interno deve iterar sobre os elementos de cada linha (com índice j)
# Imprima cada elemento no formato "m[i][j] = valor"
# Escreva seu código aqui

for i, linhas in enumerate(matriz):
  for j, coluna in enumerate(linhas):
    print(f"[{i}][{j}] = {coluna}")
````

### 1.2 Observação

O exemplo acima já cobre o fluxo típico. Se você precisar calcular médias/somas por linha, normalmente usa `sum(linha)` e `len(linha)` dentro de um `for`.

### 1.3 Modificar

Você pode alterar `m[i][j]`, adicionar linhas com `append(...)` e, se fizer sentido, adicionar colunas percorrendo as linhas (mas isso costuma indicar que talvez um `dict`/`DataFrame` seja melhor).

### 1.4 Iterar

Quando precisar de posição (linha/coluna), use `enumerate` como no exemplo único.

## **2. Compreensões de listas — o essencial**

Compreensões criam listas de forma concisa.

- **Formato base**: `[expr for item in iterável]`
- **Com filtro**: `[expr for item in iterável if condicao]`



````python
# Use uma compreensão de lista para criar uma lista de quadrados
# A lista deve conter os quadrados dos números de 0 a 4 (use range(5))
# Imprima a lista resultante
# Escreva seu código aqui

lista = []

# forma 1
for i in range(5):
  lista.append(i**2)

print(lista)

# forma 2

lista2 = [x**2 for x in range(5)]

print(lista2)
````

````python
# Crie uma lista numeros = [-2, -1, 0, 1, 2, 3]
# Use uma compreensão de lista com filtro para criar uma nova lista
# A nova lista deve conter apenas os quadrados dos números positivos (x > 0)
# Imprima a lista resultante
# Escreva seu código aqui

numeros = [-2, -1, 0, 1, 2, 3]
lista3 = [x**2 for x in range(5) if x > 0]

print(lista3)
````

````python
# Crie uma matriz = [[1, 2], [3, 4]]
# Use uma compreensão de lista aninhada para "achatar" a matriz em uma lista única
# A lista resultante deve conter todos os valores da matriz: [1, 2, 3, 4]
# Imprima a lista achatada
# Escreva seu código aqui

matriz = [[1, 2], [3, 4]]
# resultado esperado é matriz: [1, 2, 3, 4]

lista4 = [v for linha in matriz for v in linha]
print(lista4)
````

### 2.1 Observação

Se a compreensão ficar longa/difícil de ler, volte para `for` tradicional — clareza > “curto”.

### 2.2 Compreensão aninhada

Compreensão dentro de compreensão pode ser útil (ex.: criar matriz), mas use com moderação.

### 2.3 Dica de leitura

Se você ler em voz alta e ficar confuso, refatore para `for` tradicional.

## **3. Compreensões de dicionários — o essencial**

- **Formato base**: `{chave: valor for ...}`
- **Com filtro**: `{chave: valor for ... if condicao}`



````python
# Crie duas listas: nomes = ["João", "Maria", "Pedro"] e idades = [25, 30, 22]
# Use uma compreensão de dicionário com zip() para criar um dicionário
# O dicionário deve mapear cada nome à sua idade correspondente
# Imprima o dicionário resultante
# Escreva seu código aqui
nomes = ["João", "Maria", "Pedro"]
idades = [25, 30, 22]

pessoas = {nome:idade for nome, idade in zip(nomes,idades)}

print(pessoas)
````

````python
for x,y in zip(nomes, idades):
  print(x,y)
````

````python
# Crie um dicionário notas = {"João": 8.5, "Maria": 6.0, "Pedro": 9.0}
# Use uma compreensão de dicionário com filtro para criar um novo dicionário
# O novo dicionário deve conter apenas os alunos com nota >= 7 (aprovados)
# Imprima o dicionário de aprovados
# Escreva seu código aqui

notas = {"João": 8.5, "Maria": 6.0, "Pedro": 9.0}

aprovados = {nome:notas for nome, notas in notas.items() if notas >= 7}
print(aprovados)
````

### 3.1 Observação

Tal como em listas: se ficar confuso, prefira `for` normal.

- Sintaxe concisa para criar listas

- Com filtro: `[expr for item in iterável if condição]`
- Aninhadas: `[[...] for i in range(n) for j in range(m)]`

```python
quadrados = [x**2 for x in range(5)]
positivos = [x for x in numeros if x > 0]
```

### **Compreensões de Dicionários**
- Similar às listas, mas cria dicionários
- Formato: `{chave: valor for ...}`
- Com filtro: `{chave: valor for ... if condição}`

```python
dicionario = {nome: idade for nome, idade in zip(nomes, idades)}
aprovados = {nome: nota for nome, nota in notas.items() if nota >= 7}
```
