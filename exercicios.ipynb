# **EXERCÍCIOS**

```python
# Use um laço for para imprimir os números de 1 até 10.

for i in range(1, 11):
  print(i)
```

```python
# Use um laço while para contar de 1 até 5 e exibir cada número.

cont = 1
while cont <=5:
  print(f"Número: {cont}")
  cont = cont + 1
```

```python
# Use um laço for para calcular a média das notas: [7.5, 8.0, 9.5, 6.0, 8.5]

notas = [7.5, 8.0, 9.5, 6.0, 8.5]
soma = 0

for nota in notas:
  soma = soma + nota

media = soma / len(notas)

print(f"A média das notas é: {media:.2f}")
```

```python
# Use range() e for para imprimir a tabuada do 7 (de 1 até 10).
print("Tabuada do 7: ")
for i in range(1, 11):
  resultado = 7 * i

  print(f"7 x {i} = {resultado}")
```

```python
# Use range() e for para imprimir todos os números pares de 0 até 20.

print("Números pares de 0 até 20: ")

for i in range(0, 21):
  if i % 2 == 0:
    print(i)
```

```python
# Crie uma lista de números: [5, 12, 8, 3, 15, 7]. Use um laço for para encontrar o primeiro número maior que 10 e parar quando encontrar.
numeros = [5, 12, 8, 3, 15, 7]

for numero in numeros:
  if numero > 10:
    print(f"O número {numero} é o primeiro número maior que 10 da lista")
    break
else:
  print("Não há nenhum número maior do que 10 na lista")
```

```python
# Use continue para imprimir apenas os números positivos da lista: [-2, 5, -1, 8, -3, 10]

lista = [-2, 5, -1, 8, -3, 10]

for numero in lista:
  if numero < 0:
    continue
  print(numero)
```

```python
# Crie um programa que itera sobre a string "Python" e conta quantas vogais existem.

palavra = "Python"
vogais = "AEIOUaeiou"
contador_vogais = 0

for letra in palavra:
  if letra in vogais:
    contador_vogais = contador_vogais + 1

print(f"A quantidade de vogais na palavra {palavra} é: {contador_vogais}")
```

```python
# Crie um programa que itera sobre uma lista de notas: [8.5, 6.0, 9.5, 7.0, 5.5]
# e classifica cada uma (>= 9.0: Excelente, >= 7.0: Bom, >= 5.0: Regular, < 5.0: Insuficiente).

notas = [8.5, 6.0, 9.5, 7.0, 5.5]

print("CLASSIFICAÇÃO DE NOTAS: ")
for nota in notas:
  if nota >= 9.0:
    classificacao = "Excelente"
  elif nota >= 7.0:
    classificacao = "Bom"
  elif nota >= 5.0:
    classificacao = "Regular"
  else:
    classificacao = "Insuficiente"

  print(f"Nota {nota:.2f}: {classificacao}")
```
