## **1. Estruturas de dados (visão rápida)**

Em Python, as estruturas mais usadas para organizar dados são:

- **Listas (`list`)**: coleção **ordenada** e **mutável**
- **Tuplas (`tuple`)**: coleção **ordenada** e **imutável**
- **Dicionários (`dict`)**: mapeamento **chave → valor** (ordem de inserção a partir do Python 3.7)

A ideia da aula é aprender o **mínimo necessário** para criar, acessar, atualizar e percorrer cada uma delas.

## **2. Listas (`list`) — o essencial**

Use listas quando você precisa de uma sequência **ordenada** que pode **crescer/encolher**.

- **Criar**: `[]` ou `list()`
- **Acessar**: `lista[i]`, `lista[inicio:fim]`
- **Atualizar**: `append`, `extend`, `insert`, atribuição por índice
- **Remover**: `remove`, `pop`, `del`
- **Iterar**: `for item in lista` (use `enumerate` quando precisar do índice)


```python
# Exemplo único: criar, acessar, atualizar, remover e percorrer
# Crie uma lista frutas = ["maçã", "banana", "laranja"]
# Acesse o primeiro elemento e as últimas duas usando índices
# Atualize a lista: adicione "uva" com append e altere o elemento no índice 1 para "morango"
# Remova "laranja" com remove e remova o último elemento com pop
# Imprima os valores acessados e a lista final
# Use um laço for com enumerate para imprimir cada fruta numerada
# Escreva seu código aqui

# 1. Criamos uma lista

frutas = ["maçã", "banana", "laranja"]

# 2. acessar

primeiro_item = frutas[0]
print(frutas[0:3])

# 3. atualizar

frutas.append("uva")
frutas[0] = "graviola"
print(frutas)

# 4. remover

frutas.remove("uva")
print(frutas)
frutas.pop() # remove o último item
print(frutas)

# 5. percorrer

for fruta in frutas:
  print(fruta)
```

### 2.1 Acesso e fatiamento (slicing)

O básico de índice e slicing já aparece no exemplo consolidado acima. Regra prática:

- `lista[i]` acessa um item
- `lista[a:b]` pega um intervalo (não inclui `b`)
- índices negativos contam a partir do fim

### 2.2 Atualização e remoção

Use:

- `append`/`extend` para adicionar
- `remove`/`pop`/`del` para remover
- atribuição por índice para atualizar (`lista[i] = ...`)

(O exemplo consolidado já cobre esse ciclo completo.)

### 2.3 Iteração

Preferência:

- `for item in lista` quando você só precisa do valor
- `for i, item in enumerate(lista, start=1)` quando também precisa do índice

```python
lista = [5, 1, 6]
for i, item in enumerate(lista, start = 1):
  print(f"O indice {i} possui o valor {item}.")
```

### 2.4 Funções/métodos úteis (sem excesso)

- `len(lista)` tamanho
- `sum(lista)` soma (quando são números)
- `sorted(lista)` retorna **nova** lista ordenada
- `lista.sort()` ordena **no lugar**
- `lista.count(x)` conta ocorrências

```python
# Crie uma lista numeros = [3, 1, 4, 1, 5]
# Use len() para imprimir o tamanho da lista
# Use sum() para imprimir a soma dos números
# Use sorted() para imprimir uma nova lista ordenada (sem modificar a original)
# Use count(1) para contar quantas vezes o número 1 aparece
# Use sort() para ordenar a lista no lugar e imprima o resultado
# Escreva seu código aqui

numeros = [3, 1, 4, 1, 5]

# 1. Tamanho

print(f"O tamanho da lista é {len(numeros)}")

# 2. Soma

print(f"A soma de todos os valores da lista é {sum(numeros)}")

# 3. Ordenar

print(f"A lista ordenada é: {sorted(numeros)}")

# 4. Contar ocorrencias

print(f"A lista possúi {numeros.count(1)} vezes o número 1")

# 5. Ordenar com sort
numeros.sort() # A estrutura da lista é alterada
print(numeros)
```

## **3. Tuplas (`tuple`) — quando você quer imutabilidade**

Tuplas são como listas, mas **não devem ser alteradas**.

Use quando:
- o conjunto de valores é “fixo” (ex.: coordenadas)
- você quer **garantia** de que não vai mudar sem querer

Acesso por índice e iteração funcionam igual às listas.

```python
# Crie uma tupla coordenadas = (10, 20)
# Desempacote a tupla em variáveis x e y
# Imprima os valores de x e y
# Tente modificar coordenadas[0] = 99 dentro de um bloco try/except
# Capture o TypeError e imprima uma mensagem explicando que tuplas são imutáveis
# Escreva seu código aqui

coordenadas = (10, 20)

# 1. Imprimir e atribuir os valores da tupla
x,y = coordenadas
print(x,y)

```

```python
# 2. Tentativa de modificação

try:
  coordenadas[0] = 99
except:
  print("Você não pode alterar os valores dos itens da tupla")
```

```python
# 3. Captura de erro
try:
  coordenadas[0] = 99
except TypeError as e:
  print("Você não pode alterar os valores dos itens da tupla")
  print(e)
```

### 3.1 Acessar / desempacotar

- `tupla[i]` acessa um item
- `a, b = tupla` desempacota (quantidade deve bater)

O exemplo acima já cobre isso.

### 3.2 Iteração

Mesma ideia de lista:

- `for item in tupla`
- `for i, item in enumerate(tupla)`

### 3.3 Quando usar?

- dados fixos (ex.: coordenadas)
- retornos múltiplos de função
- “registro” simples onde a posição tem significado

## **4. Dicionários (`dict`) — chave → valor**

Use dicionários quando você quer acessar dados por uma **chave** (nomeada), em vez de índice.

- **Criar**: `{}` ou `dict()`
- **Acessar**: `d[chave]` (exige existir) ou `d.get(chave, padrão)`
- **Atualizar**: `d[chave] = valor`
- **Iterar**: `for chave, valor in d.items()`

```python
# Crie um dicionário pessoa = {"nome": "João", "idade": 25}
# Acesse o valor da chave "nome" usando pessoa["nome"] e imprima
# Acesse a chave "telefone" usando get() com valor padrão "(não informado)" e imprima
# Adicione a chave "cidade" com valor "São Paulo"
# Atualize a idade incrementando em 1
# Use um laço for com items() para iterar e imprimir cada chave e valor
# Remova a chave "cidade" usando pop() e imprima o dicionário final
# Escreva seu código aqui

pessoa = {"nome": "João", "idade": 25}

# 1. Acessar o valor da chave nome
print(pessoa["nome"])

# 2. Acesso com get
print(pessoa.get("telefone", "valor não informado"))

# 3. Adicionar chave

pessoa["cidade"] = "SP"
print(pessoa)

# 4. Atualizar valor de idade
pessoa["idade"] = pessoa["idade"] + 1
print(pessoa["idade"])

# 5. Iterar e imprimir valores
for chave, valor in pessoa.items():
  print(chave,"->", valor)

# 6. Remover ultimo valor
pessoa.pop("cidade") # Você precisa referenciar a chave
print(pessoa)
```

### 4.1 Acesso seguro

- `d[chave]` é ótimo quando a chave **existe**
- `d.get(chave, padrão)` evita `KeyError` quando pode faltar

### 4.2 Atualizar e remover

- `d[chave] = valor` adiciona/atualiza
- `pop(chave)` remove e retorna o valor
- `del d[chave]` remove (sem retorno)

### 4.3 Iterar (a forma mais comum)

Use `items()` quando quiser chave e valor:

- `for chave, valor in d.items(): ...`

## **5. Comparação rápida**

| Característica | `list` | `tuple` | `dict` |
|---|---:|---:|---:|
| Mantém ordem? | ✅ | ✅ | ✅ (ordem de inserção) |
| Mutável? | ✅ | ❌ | ✅ |
| Acesso principal | índice | índice | chave |
| Duplicatas | ✅ | ✅ | ✅ (valores) / ❌ (chaves) |

## **Resumo da Aula 1**

Nesta aula, você aprendeu sobre as três principais estruturas de dados em Python:

### **Listas (`list`)**
- **Mutáveis** e **ordenadas**
- Usar quando precisa de sequências que podem crescer/encolher
- Operações: `append()`, `remove()`, `pop()`, `insert()`, `extend()`
- Acesso por índice: `lista[i]`, `lista[a:b]` (slicing)

### **Tuplas (`tuple`)**
- **Imutáveis** e **ordenadas**
- Usar quando precisa de dados fixos que não devem mudar
- Acesso igual às listas, mas não pode modificar elementos
- Útil para desempacotamento: `x, y = coordenadas`

### **Dicionários (`dict`)**
- **Mutáveis** e **ordenados** (ordem de inserção)
- Usar quando precisa acessar dados por **chave** (nomeada)
- Operações: `d[chave] = valor`, `d.get(chave, padrão)`, `pop()`
- Iteração comum: `for chave, valor in d.items()`

### **Escolha a estrutura certa:**
- **Lista**: sequências mutáveis (ex: lista de compras)
- **Tupla**: dados fixos (ex: coordenadas)
- **Dicionário**: acesso por nome/chave (ex: cadastro de pessoas)

```python
# Dicionários — desafio
# Você tem uma lista de dicionários representando alunos:

alunos = [
    {"nome": "Ana", "nota": 8.5},
    {"nome": "Bruno", "nota": 6.0},
    {"nome": "Carla", "nota": 9.2}
]

#Imprima o nome de cada aluno com sua nota
#Calcule a média das notas da turma
#Diga quais alunos tiveram nota acima da média

for aluno in alunos:
  print(aluno["nome"], aluno["nota"])


soma_notas = 0
for aluno in alunos:
  soma_notas = soma_notas + aluno["nota"]

media = soma_notas / len(alunos)
print(f"A média da turma é: {media:.2f}")

for aluno in alunos:
  if aluno["nota"] > media:
    print(f"O aluno(a) {aluno["nome"]} esta acima da média")
```
